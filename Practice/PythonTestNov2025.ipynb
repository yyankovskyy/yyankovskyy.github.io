```python
import pandas as pd

# Sample data
data = {
    'product_category': ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C'],
    'product_id': ['101', '102', '103', '201', '202', '203', '301', '302', '303'],
    'sales': [100, 200, 150, 300, 250, 200, 400, 500, 300]
}
df = pd.DataFrame(data)
df.head(3)
```

```python
# Calculate total sales per product within each category
df_grouped = df.groupby(['product_category', 'product_id']).agg(total_sales=('sales', 'sum')).reset_index()
df_grouped.head(3)
```

```python
df_grouped = df.groupby(['product_category', 'product_id'])['sales'].agg('sum').reset_index()
df.group
df_grouped.head(3)
```

```python
# Calculate total sales per category
category_totals = df.groupby('product_category')['sales'].sum().rename('category_total_sales')
category_totals.head(3)
```

```python
category_totals = df.groupby('product_category').agg(category_total_sales=('sales', 'sum')).reset_index()
category_totals.head(3)
```

```python
# Merge total sales per product with total sales per category
df_merged = pd.merge(df_grouped, category_totals, on='product_category')

# Calculate the sales percentage of each product within its category
df_merged['sales_percentage'] = (df_merged['total_sales'] / df_merged['category_total_sales']) * 100

# Display the result
print(df_merged[['product_category', 'product_id', 'sales_percentage']])
```

```python
data = {
    'transaction_date': ['2022-01-15', '2023-01-20', '2022-02-15', '2023-02-20', '2022-03-15', '2023-04-01'],
    'product_id': ['101', '101', '102', '102', '101', '102'],
    'sales': [100, 150, 200, 250, 300, 350]
}
df = pd.DataFrame(data)
```

```python
df['month'] = pd.to_datetime(df['transaction_date']).dt.to_period('M')
monthly_sales = df.groupby(['product_id', 'month'])['sales'].sum().unstack().fillna(0)
monthly_sales.head(2)
```
